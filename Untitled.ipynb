```python
current_date = df['order_date'].max() + dt.timedelta(days=1)

rfm = df.groupby('customer_name').agg({
    'order_date': lambda x: (current_date - x.max()).days, # Recency
    'order_id': 'count',
    # Frequency
    'profit': 'sum'
    # Monetary
})

rfm.columns = ['Recency','Frequency','Monetary']

rfm['R_score'] = pd.qcut(rfm['Recency'], 5, labels=[5,4,3,2,1])
rfm['F_score'] = pd.qcut(rfm['Frequency'], 5, labels=[1,2,3,4,5])
rfm['M_score'] = pd.qcut(rfm['Monetary'], 5, labels=[1,2,3,4,5])

rfm['RFM_Score'] = rfm['R_score'].astype(str) + rfm['F_score'].astype(str) + rfm['M_score'].astype(str)

def segment_customer(df):
    if df['RFM_Score']=='555':
        return "Champions"
    elif df['R_score'] >= 4 and df['F_score'] >= 4:
        return "Loyal Customers"
    elif df['R_score'] <= 2:
        return "At Risk"
    else:
        return "Regulars"

rfm['Segment'] = rfm.apply(segment_customer, axis=1)

print(rfm.head())


rfm['Segment'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(8,8))
plt.title('Customer Segment Distribution')
plt.show()


                        
                
                         
```

### Advanced Customer Segmentation (RFM Analysis)

To move beyond basic sales tracking, I implemented an **RFM (Recency, Frequency, Monetary)** model to evaluate customer behavior and lifetime value.

#### **Key Findings:**
* **Champions (2.0%):** Our most valuable customers. They buy recently, frequently, and spend the most.
  * *Strategy:* Reward with VIP loyalty programs and early access to new products.
* **Loyal Customers (16.2%):** These are our most reliable recurring revenue sources. They have a high putchase frequency and a proven affinity for the brand.
  * *Strategy:* They dont need heavy discount to buy. Instead, focus on **retention**; offer them "Referral Bonuses" or "Product Bundles" to increase their basket size.
* **At Risk(39.2%):** A significant portion of the customer base has not purchased recently.
  * *Strategy:* Immediate "Win-back" email campaigns and personalized discounts are required to prevent permanent churn.
* **Regulars (42.5%):** The backbone of the business.
  * *Strategy:* Focus on "Upselling" higher margin technology products to increase their average order value.

**Insight:** By identifying that nearly **40% of customer are 'At risk'**, we can we can shift from a reactive to a proactive marketing strategy, potentially saving thousands in lost future revenue.

```python
import scipy.stats as stats
import seaborn as sns

tech_profit = df[df['category']=='Technology']['profit']
office_profit = df[df['category']=='Office Supplies']['profit']
furniture_profit = df[df['category']=='Furniture']['profit']

f_stat, p_value = stats.f_oneway(tech_profit, office_profit, furniture_profit)

plt.figure(figsize=(10,6))
sns.boxplot(x='category', y='profit', data=df)
plt.ylim(-200,400)
plt.title('Profit Distribution by Category (Statistical Comparison)')
plt.show()

print(f"F-Statistic: {f_stat:.2f}")
print(f"p-Value:{p_value:.4f}")

if p_value < 0.05:
    print("Result: Significant diffrenece found between categories (Reject Null Hypothesis)")
else: 
    print("Result: no significant difference found (Fail to reject Null Hypothesis)")
    
    
    
```

### Statistical Validation & Logistics Optimization

To ensure that my insight were based on random fluctuations, I continiued high-level statistical testic and logistical cost analysis.

#### **1. ANOVA Test: Categorical Profitability Validation**
* **Finding:** The **P-value (0.0000)** and **F-statistic (304.23)** confirm a statistically significant difference in profit margins across categories.
*  **Insight:** We have a 99.9% confidence level that **Technology** is inherently more profitable than furniture and office supplies. This justifies a strategic shift in marketing budget allocation toward high-margin tech products.

#### **2. Logistics and Shipping Cost Analysis**
* **Standard Class Efficiency:** Although it accounts for the highest volume  (**30,755 orders**), it maintains the lowest average cost (**$19.97**).
* **Premium Leakage:** **Same Day** shiping is **115% more expensive** than Standard Class but is used in less than 7% of total orders.
* **Recommendation:** Since Standard Class is the most stable and cost effective, the business should maintain its current shipping mix while investing if "First Class" costs can be negotiated down to "Second Class" levels (-$30).
